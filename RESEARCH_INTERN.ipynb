````python
import torch.nn.functional as F

class DoubleConv(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1),
            nn.BatchNorm2d(out_ch),
            nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1),
            nn.BatchNorm2d(out_ch),
            nn.ReLU(inplace=True)
        )
    def forward(self, x):
        return self.conv(x)

class UNet(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1 = DoubleConv(in_ch, base)
        self.enc2 = DoubleConv(base, base*2)
        self.enc3 = DoubleConv(base*2, base*4)
        self.pool = nn.MaxPool2d(2)

        self.bottleneck = DoubleConv(base*4, base*8)

        self.up3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2)
        self.dec3 = DoubleConv(base*8, base*4)
        self.up2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2)
        self.dec2 = DoubleConv(base*4, base*2)
        self.up1 = nn.ConvTranspose2d(base*2, base, 2, stride=2)
        self.dec1 = DoubleConv(base*2, base)

        self.out_conv = nn.Conv2d(base, out_ch, 1)

    def forward(self, x):
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))

        d3 = self.up3(b)
        d3 = self.dec3(torch.cat([d3, e3], dim=1))
        d2 = self.up2(d3)
        d2 = self.dec2(torch.cat([d2, e2], dim=1))
        d1 = self.up1(d2)
        d1 = self.dec1(torch.cat([d1, e1], dim=1))

        return self.out_conv(d1)

# Sanity check: forward pass on real data
model = UNet()
test_input = sparse_tensor[:2]  # small batch of 2
with torch.no_grad():
    output = model(test_input)
print("Input shape:", test_input.shape)
print("Output shape:", output.shape)
````

````python
import numpy as np
from skimage.transform import radon, iradon
from PIL import Image
import matplotlib.pyplot as plt

def simulate_sparse_view(gt_image, num_angles=32, full_angles=180):
    """
    gt_image: 2D numpy array (clean ground truth)
    num_angles: number of projection angles used (fewer = sparser/more artifacts)
    full_angles: total angle range simulated (standard = 180)
    """
    theta_full = np.linspace(0., full_angles, full_angles, endpoint=False)
    theta_sparse = np.linspace(0., full_angles, num_angles, endpoint=False)

    # Forward: simulate sparse sinogram (only num_angles projections)
    sinogram_sparse = radon(gt_image, theta=theta_sparse, circle=True)

    # Inverse: reconstruct from sparse data -> produces artefacts
    sparse_recon = iradon(sinogram_sparse, theta=theta_sparse, circle=True)

    return sparse_recon

# Test on one image
gt = np.array(Image.open("/content/vessel_data/Vessel/train/0.tif")).astype(np.float64)
sparse = simulate_sparse_view(gt, num_angles=32)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(sparse, cmap='gray')
axes[0].set_title('Simulated sparse (32 angles)')
axes[1].imshow(gt, cmap='gray')
axes[1].set_title('Ground truth')
plt.show()
````

````python
from torch.utils.data import Dataset, DataLoader

class PADataset(Dataset):
    def __init__(self, sparse_dir, gt_dir):
        self.sparse_dir = sparse_dir
        self.gt_dir = gt_dir
        self.files = sorted([f for f in os.listdir(sparse_dir) if f.endswith('.tif')])

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        fname = self.files[idx]
        sparse = np.array(Image.open(os.path.join(self.sparse_dir, fname))).astype(np.float32) / 255.0
        gt = np.array(Image.open(os.path.join(self.gt_dir, fname))).astype(np.float32) / 255.0

        sparse = torch.from_numpy(sparse).unsqueeze(0)  # add channel dim -> (1, H, W)
        gt = torch.from_numpy(gt).unsqueeze(0)

        return sparse, gt

# Build datasets for the 32-angle sparsity level
train_ds = PADataset("/content/vessel_data/Vessel_sparse32/train", "/content/vessel_data/Vessel/train")
test_ds = PADataset("/content/vessel_data/Vessel_sparse32/test", "/content/vessel_data/Vessel/test")
val_ds = PADataset("/content/vessel_data/Vessel_sparse32/validation", "/content/vessel_data/Vessel/validation")

train_loader = DataLoader(train_ds, batch_size=16, shuffle=True)
test_loader = DataLoader(test_ds, batch_size=16, shuffle=False)
val_loader = DataLoader(val_ds, batch_size=16, shuffle=False)

# Sanity check
sparse_batch, gt_batch = next(iter(train_loader))
print("Sparse batch shape:", sparse_batch.shape)
print("GT batch shape:", gt_batch.shape)
````

````python
import torch.optim as optim
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

model = UNet(in_ch=1, out_ch=1, base=32).to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=1e-3)

num_epochs = 20

for epoch in range(num_epochs):
    model.train()
    train_loss = 0.0
    for sparse, gt in train_loader:
        sparse, gt = sparse.to(device), gt.to(device)

        optimizer.zero_grad()
        output = model(sparse)
        loss = criterion(output, gt)
        loss.backward()
        optimizer.step()

        train_loss += loss.item() * sparse.size(0)

    train_loss /= len(train_ds)

    # Quick validation pass
    model.eval()
    val_loss = 0.0
    with torch.no_grad():
        for sparse, gt in val_loader:
            sparse, gt = sparse.to(device), gt.to(device)
            output = model(sparse)
            loss = criterion(output, gt)
            val_loss += loss.item() * sparse.size(0)
    val_loss /= len(val_ds)

    print(f"Epoch {epoch+1}/{num_epochs} | Train Loss: {train_loss:.5f} | Val Loss: {val_loss:.5f}")
````

````python
class SEBlock(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(
            nn.Linear(channels, channels // reduction, bias=False),
            nn.ReLU(inplace=True),
            nn.Linear(channels // reduction, channels, bias=False),
            nn.Sigmoid()
        )

    def forward(self, x):
        b, c, _, _ = x.shape
        y = self.avg_pool(x).view(b, c)
        y = self.fc(y).view(b, c, 1, 1)
        return x * y.expand_as(x)
````

````python
class DoubleConvSE(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1),
            nn.BatchNorm2d(out_ch),
            nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1),
            nn.BatchNorm2d(out_ch),
            nn.ReLU(inplace=True)
        )
        self.se = SEBlock(out_ch)

    def forward(self, x):
        x = self.conv(x)
        return self.se(x)

class UNet_SE(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1 = DoubleConvSE(in_ch, base)
        self.enc2 = DoubleConvSE(base, base*2)
        self.enc3 = DoubleConvSE(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvSE(base*4, base*8)
        self.up3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2)
        self.dec3 = DoubleConvSE(base*8, base*4)
        self.up2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2)
        self.dec2 = DoubleConvSE(base*4, base*2)
        self.up1 = nn.ConvTranspose2d(base*2, base, 2, stride=2)
        self.dec1 = DoubleConvSE(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)

    def forward(self, x):
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.up3(b)
        d3 = self.dec3(torch.cat([d3, e3], dim=1))
        d2 = self.up2(d3)
        d2 = self.dec2(torch.cat([d2, e2], dim=1))
        d1 = self.up1(d2)
        d1 = self.dec1(torch.cat([d1, e1], dim=1))
        return self.out_conv(d1)

# Sanity check
model_se = UNet_SE().to(device)
test_out = model_se(sparse_sample.unsqueeze(0).to(device))
print("SE U-Net output shape:", test_out.shape)
````

````python
class ChannelAttention(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.max_pool = nn.AdaptiveMaxPool2d(1)
        self.fc = nn.Sequential(
            nn.Linear(channels, channels // reduction, bias=False),
            nn.ReLU(inplace=True),
            nn.Linear(channels // reduction, channels, bias=False)
        )
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        b, c, _, _ = x.shape
        avg_out = self.fc(self.avg_pool(x).view(b, c))
        max_out = self.fc(self.max_pool(x).view(b, c))
        out = self.sigmoid(avg_out + max_out).view(b, c, 1, 1)
        return x * out

class SpatialAttention(nn.Module):
    def __init__(self, kernel_size=7):
        super().__init__()
        self.conv = nn.Conv2d(2, 1, kernel_size, padding=kernel_size//2, bias=False)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        avg_out = torch.mean(x, dim=1, keepdim=True)
        max_out, _ = torch.max(x, dim=1, keepdim=True)
        out = torch.cat([avg_out, max_out], dim=1)
        out = self.sigmoid(self.conv(out))
        return x * out

class CBAM(nn.Module):
    def __init__(self, channels, reduction=16, kernel_size=7):
        super().__init__()
        self.ca = ChannelAttention(channels, reduction)
        self.sa = SpatialAttention(kernel_size)

    def forward(self, x):
        x = self.ca(x)
        x = self.sa(x)
        return x
````

````python
class SimAM(nn.Module):
    def __init__(self, e_lambda=1e-4):
        super().__init__()
        self.e_lambda = e_lambda
        self.activation = nn.Sigmoid()

    def forward(self, x):
        b, c, h, w = x.shape
        n = w * h - 1
        x_minus_mu_square = (x - x.mean(dim=[2, 3], keepdim=True)).pow(2)
        y = x_minus_mu_square / (4 * (x_minus_mu_square.sum(dim=[2, 3], keepdim=True) / n + self.e_lambda)) + 0.5
        return x * self.activation(y)
````

````python
# --- Rebuild baseline U-Net architecture ---
class DoubleConv(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True)
        )
    def forward(self, x):
        return self.conv(x)

class UNet(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1 = DoubleConv(in_ch, base)
        self.enc2 = DoubleConv(base, base*2)
        self.enc3 = DoubleConv(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConv(base*4, base*8)
        self.up3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2)
        self.dec3 = DoubleConv(base*8, base*4)
        self.up2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2)
        self.dec2 = DoubleConv(base*4, base*2)
        self.up1 = nn.ConvTranspose2d(base*2, base, 2, stride=2)
        self.dec1 = DoubleConv(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

# --- Rebuild SE architecture too, since we'll need it for param counts ---
class SEBlock(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(
            nn.Linear(channels, channels // reduction, bias=False), nn.ReLU(inplace=True),
            nn.Linear(channels // reduction, channels, bias=False), nn.Sigmoid()
        )
    def forward(self, x):
        b, c, _, _ = x.shape
        y = self.avg_pool(x).view(b, c)
        y = self.fc(y).view(b, c, 1, 1)
        return x * y.expand_as(x)

class DoubleConvSE(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True)
        )
        self.se = SEBlock(out_ch)
    def forward(self, x):
        return self.se(self.conv(x))

class UNet_SE(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1 = DoubleConvSE(in_ch, base)
        self.enc2 = DoubleConvSE(base, base*2)
        self.enc3 = DoubleConvSE(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvSE(base*4, base*8)
        self.up3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2)
        self.dec3 = DoubleConvSE(base*8, base*4)
        self.up2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2)
        self.dec2 = DoubleConvSE(base*4, base*2)
        self.up1 = nn.ConvTranspose2d(base*2, base, 2, stride=2)
        self.dec1 = DoubleConvSE(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

# --- Load checkpoints ---
model = UNet(in_ch=1, out_ch=1, base=32).to(device)
model.load_state_dict(torch.load("/content/drive/MyDrive/PA_project/checkpoints/baseline_unet_sparse32.pth"))

model_se = UNet_SE(in_ch=1, out_ch=1, base=32).to(device)
model_se.load_state_dict(torch.load("/content/drive/MyDrive/PA_project/checkpoints/se_unet_sparse32.pth"))

print("Baseline and SE models restored from checkpoints.")
````

````python
import os
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from PIL import Image
from torch.utils.data import Dataset, DataLoader
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# --- Baseline U-Net ---
class DoubleConv(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True)
        )
    def forward(self, x): return self.conv(x)

class UNet(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConv(in_ch, base), DoubleConv(base, base*2), DoubleConv(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConv(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConv(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConv(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConv(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

# --- SE ---
class SEBlock(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(nn.Linear(channels, channels//reduction, bias=False), nn.ReLU(inplace=True),
                                 nn.Linear(channels//reduction, channels, bias=False), nn.Sigmoid())
    def forward(self, x):
        b, c, _, _ = x.shape
        y = self.fc(self.avg_pool(x).view(b, c)).view(b, c, 1, 1)
        return x * y.expand_as(x)

class DoubleConvSE(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True)
        )
        self.se = SEBlock(out_ch)
    def forward(self, x): return self.se(self.conv(x))

class UNet_SE(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConvSE(in_ch, base), DoubleConvSE(base, base*2), DoubleConvSE(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvSE(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConvSE(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConvSE(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConvSE(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

# --- CBAM ---
class ChannelAttention(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        self.avg_pool, self.max_pool = nn.AdaptiveAvgPool2d(1), nn.AdaptiveMaxPool2d(1)
        self.fc = nn.Sequential(nn.Linear(channels, channels//reduction, bias=False), nn.ReLU(inplace=True),
                                 nn.Linear(channels//reduction, channels, bias=False))
        self.sigmoid = nn.Sigmoid()
    def forward(self, x):
        b, c, _, _ = x.shape
        avg_out = self.fc(self.avg_pool(x).view(b, c))
        max_out = self.fc(self.max_pool(x).view(b, c))
        return x * self.sigmoid(avg_out + max_out).view(b, c, 1, 1)

class SpatialAttention(nn.Module):
    def __init__(self, kernel_size=7):
        super().__init__()
        self.conv = nn.Conv2d(2, 1, kernel_size, padding=kernel_size//2, bias=False)
        self.sigmoid = nn.Sigmoid()
    def forward(self, x):
        avg_out = torch.mean(x, dim=1, keepdim=True)
        max_out, _ = torch.max(x, dim=1, keepdim=True)
        return x * self.sigmoid(self.conv(torch.cat([avg_out, max_out], dim=1)))

class CBAM(nn.Module):
    def __init__(self, channels, reduction=16, kernel_size=7):
        super().__init__()
        self.ca, self.sa = ChannelAttention(channels, reduction), SpatialAttention(kernel_size)
    def forward(self, x): return self.sa(self.ca(x))

class DoubleConvCBAM(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True)
        )
        self.cbam = CBAM(out_ch)
    def forward(self, x): return self.cbam(self.conv(x))

class UNet_CBAM(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConvCBAM(in_ch, base), DoubleConvCBAM(base, base*2), DoubleConvCBAM(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvCBAM(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConvCBAM(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConvCBAM(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConvCBAM(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

# --- SimAM ---
class SimAM(nn.Module):
    def __init__(self, e_lambda=1e-4):
        super().__init__()
        self.e_lambda = e_lambda
        self.activation = nn.Sigmoid()
    def forward(self, x):
        b, c, h, w = x.shape
        n = w * h - 1
        x_minus_mu_square = (x - x.mean(dim=[2, 3], keepdim=True)).pow(2)
        y = x_minus_mu_square / (4 * (x_minus_mu_square.sum(dim=[2, 3], keepdim=True) / n + self.e_lambda)) + 0.5
        return x * self.activation(y)

class DoubleConvSimAM(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True)
        )
        self.simam = SimAM()
    def forward(self, x): return self.simam(self.conv(x))

class UNet_SimAM(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConvSimAM(in_ch, base), DoubleConvSimAM(base, base*2), DoubleConvSimAM(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvSimAM(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConvSimAM(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConvSimAM(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConvSimAM(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

# --- Dataset class (needed since fresh session) ---
class PADataset(Dataset):
    def __init__(self, sparse_dir, gt_dir):
        self.sparse_dir, self.gt_dir = sparse_dir, gt_dir
        self.files = sorted([f for f in os.listdir(sparse_dir) if f.endswith('.tif')])
    def __len__(self): return len(self.files)
    def __getitem__(self, idx):
        fname = self.files[idx]
        sparse = np.array(Image.open(os.path.join(self.sparse_dir, fname))).astype(np.float32) / 255.0
        gt = np.array(Image.open(os.path.join(self.gt_dir, fname))).astype(np.float32) / 255.0
        return torch.from_numpy(sparse).unsqueeze(0), torch.from_numpy(gt).unsqueeze(0)

# --- Reusable training + eval function ---
def train_and_eval(model_class, name, train_loader, val_loader, test_loader, train_ds, val_ds, num_epochs=20):
    model = model_class(in_ch=1, out_ch=1, base=32).to(device)
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=1e-3)

    for epoch in range(num_epochs):
        model.train()
        train_loss = 0.0
        for sparse, gt in train_loader:
            sparse, gt = sparse.to(device), gt.to(device)
            optimizer.zero_grad()
            output = model(sparse)
            loss = criterion(output, gt)
            loss.backward()
            optimizer.step()
            train_loss += loss.item() * sparse.size(0)
        train_loss /= len(train_ds)

        model.eval()
        val_loss = 0.0
        with torch.no_grad():
            for sparse, gt in val_loader:
                sparse, gt = sparse.to(device), gt.to(device)
                output = model(sparse)
                val_loss += criterion(output, gt).item() * sparse.size(0)
        val_loss /= len(val_ds)
        print(f"[{name}] Epoch {epoch+1}/{num_epochs} | Train: {train_loss:.5f} | Val: {val_loss:.5f}")

    torch.save(model.state_dict(), f"/content/drive/MyDrive/PA_project/checkpoints/{name}.pth")

    model.eval()
    psnr_scores, ssim_scores = [], []
    with torch.no_grad():
        for sparse, gt in test_loader:
            sparse, gt = sparse.to(device), gt.to(device)
            output = model(sparse)
            out_np, gt_np = output.cpu().numpy(), gt.cpu().numpy()
            for i in range(out_np.shape[0]):
                psnr_scores.append(psnr(gt_np[i,0], out_np[i,0], data_range=1.0))
                ssim_scores.append(ssim(gt_np[i,0], out_np[i,0], data_range=1.0))

    p, s = np.mean(psnr_scores), np.mean(ssim_scores)
    print(f"{name} — Test PSNR: {p:.2f} dB | Test SSIM: {s:.4f}")
    return model, p, s

print("All architectures and training function rebuilt.")
````

````python
import os
import numpy as np
import torch
import torch.nn as nn
from PIL import Image
from torch.utils.data import Dataset
import matplotlib.pyplot as plt

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# --- Mount Drive if not already ---
from google.colab import drive
drive.mount('/content/drive')

# --- Restore data if not already present ---
if not os.path.exists("/content/vessel_data"):
    !cp -r /content/drive/MyDrive/PA_project/vessel_data /content/vessel_data

# --- Dataset class ---
class PADataset(Dataset):
    def __init__(self, sparse_dir, gt_dir):
        self.sparse_dir, self.gt_dir = sparse_dir, gt_dir
        self.files = sorted([f for f in os.listdir(sparse_dir) if f.endswith('.tif')])
    def __len__(self): return len(self.files)
    def __getitem__(self, idx):
        fname = self.files[idx]
        sparse = np.array(Image.open(os.path.join(self.sparse_dir, fname))).astype(np.float32) / 255.0
        gt = np.array(Image.open(os.path.join(self.gt_dir, fname))).astype(np.float32) / 255.0
        return torch.from_numpy(sparse).unsqueeze(0), torch.from_numpy(gt).unsqueeze(0)

test_ds = PADataset("/content/vessel_data/Vessel_sparse32/test", "/content/vessel_data/Vessel/test")

# --- Architectures ---
class DoubleConv(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
                                   nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True))
    def forward(self, x): return self.conv(x)

class UNet(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConv(in_ch, base), DoubleConv(base, base*2), DoubleConv(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConv(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConv(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConv(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConv(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

class SEBlock(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(nn.Linear(channels, channels//reduction, bias=False), nn.ReLU(inplace=True),
                                 nn.Linear(channels//reduction, channels, bias=False), nn.Sigmoid())
    def forward(self, x):
        b, c, _, _ = x.shape
        y = self.fc(self.avg_pool(x).view(b, c)).view(b, c, 1, 1)
        return x * y.expand_as(x)

class DoubleConvSE(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
                                   nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True))
        self.se = SEBlock(out_ch)
    def forward(self, x): return self.se(self.conv(x))

class UNet_SE(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConvSE(in_ch, base), DoubleConvSE(base, base*2), DoubleConvSE(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvSE(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConvSE(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConvSE(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConvSE(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

class ChannelAttention(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        self.avg_pool, self.max_pool = nn.AdaptiveAvgPool2d(1), nn.AdaptiveMaxPool2d(1)
        self.fc = nn.Sequential(nn.Linear(channels, channels//reduction, bias=False), nn.ReLU(inplace=True),
                                 nn.Linear(channels//reduction, channels, bias=False))
        self.sigmoid = nn.Sigmoid()
    def forward(self, x):
        b, c, _, _ = x.shape
        avg_out = self.fc(self.avg_pool(x).view(b, c))
        max_out = self.fc(self.max_pool(x).view(b, c))
        return x * self.sigmoid(avg_out + max_out).view(b, c, 1, 1)

class SpatialAttention(nn.Module):
    def __init__(self, kernel_size=7):
        super().__init__()
        self.conv = nn.Conv2d(2, 1, kernel_size, padding=kernel_size//2, bias=False)
        self.sigmoid = nn.Sigmoid()
    def forward(self, x):
        avg_out = torch.mean(x, dim=1, keepdim=True)
        max_out, _ = torch.max(x, dim=1, keepdim=True)
        return x * self.sigmoid(self.conv(torch.cat([avg_out, max_out], dim=1)))

class CBAM(nn.Module):
    def __init__(self, channels, reduction=16, kernel_size=7):
        super().__init__()
        self.ca, self.sa = ChannelAttention(channels, reduction), SpatialAttention(kernel_size)
    def forward(self, x): return self.sa(self.ca(x))

class DoubleConvCBAM(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
                                   nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True))
        self.cbam = CBAM(out_ch)
    def forward(self, x): return self.cbam(self.conv(x))

class UNet_CBAM(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConvCBAM(in_ch, base), DoubleConvCBAM(base, base*2), DoubleConvCBAM(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvCBAM(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConvCBAM(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConvCBAM(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConvCBAM(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

class SimAM(nn.Module):
    def __init__(self, e_lambda=1e-4):
        super().__init__()
        self.e_lambda = e_lambda
        self.activation = nn.Sigmoid()
    def forward(self, x):
        b, c, h, w = x.shape
        n = w * h - 1
        x_minus_mu_square = (x - x.mean(dim=[2, 3], keepdim=True)).pow(2)
        y = x_minus_mu_square / (4 * (x_minus_mu_square.sum(dim=[2, 3], keepdim=True) / n + self.e_lambda)) + 0.5
        return x * self.activation(y)

class DoubleConvSimAM(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
                                   nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True))
        self.simam = SimAM()
    def forward(self, x): return self.simam(self.conv(x))

class UNet_SimAM(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConvSimAM(in_ch, base), DoubleConvSimAM(base, base*2), DoubleConvSimAM(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvSimAM(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConvSimAM(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConvSimAM(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConvSimAM(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

# --- Load all 4 trained checkpoints ---
ckpt_dir = "/content/drive/MyDrive/PA_project/checkpoints"

model = UNet().to(device)
model.load_state_dict(torch.load(f"{ckpt_dir}/baseline_unet_sparse32.pth"))

model_se = UNet_SE().to(device)
model_se.load_state_dict(torch.load(f"{ckpt_dir}/se_unet_sparse32.pth"))

model_cbam = UNet_CBAM().to(device)
model_cbam.load_state_dict(torch.load(f"{ckpt_dir}/cbam_unet_sparse32.pth"))

model_simam = UNet_SimAM().to(device)
model_simam.load_state_dict(torch.load(f"{ckpt_dir}/simam_unet_sparse32.pth"))

print("All 4 models restored and ready.")
````

````python
setup_code = '''
import os
import numpy as np
import torch
import torch.nn as nn
from PIL import Image
from torch.utils.data import Dataset, DataLoader
import torch.optim as optim
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim
import matplotlib.pyplot as plt

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

if not os.path.exists("/content/vessel_data"):
    os.system("cp -r /content/drive/MyDrive/PA_project/vessel_data /content/vessel_data")

class PADataset(Dataset):
    def __init__(self, sparse_dir, gt_dir):
        self.sparse_dir, self.gt_dir = sparse_dir, gt_dir
        self.files = sorted([f for f in os.listdir(sparse_dir) if f.endswith(".tif")])
    def __len__(self): return len(self.files)
    def __getitem__(self, idx):
        fname = self.files[idx]
        sparse = np.array(Image.open(os.path.join(self.sparse_dir, fname))).astype(np.float32) / 255.0
        gt = np.array(Image.open(os.path.join(self.gt_dir, fname))).astype(np.float32) / 255.0
        return torch.from_numpy(sparse).unsqueeze(0), torch.from_numpy(gt).unsqueeze(0)

class DoubleConv(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
                                   nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True))
    def forward(self, x): return self.conv(x)

class UNet(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConv(in_ch, base), DoubleConv(base, base*2), DoubleConv(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConv(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConv(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConv(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConv(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

class SEBlock(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(nn.Linear(channels, channels//reduction, bias=False), nn.ReLU(inplace=True),
                                 nn.Linear(channels//reduction, channels, bias=False), nn.Sigmoid())
    def forward(self, x):
        b, c, _, _ = x.shape
        y = self.fc(self.avg_pool(x).view(b, c)).view(b, c, 1, 1)
        return x * y.expand_as(x)

class DoubleConvSE(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
                                   nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True))
        self.se = SEBlock(out_ch)
    def forward(self, x): return self.se(self.conv(x))

class UNet_SE(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConvSE(in_ch, base), DoubleConvSE(base, base*2), DoubleConvSE(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvSE(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConvSE(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConvSE(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConvSE(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

class ChannelAttention(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        self.avg_pool, self.max_pool = nn.AdaptiveAvgPool2d(1), nn.AdaptiveMaxPool2d(1)
        self.fc = nn.Sequential(nn.Linear(channels, channels//reduction, bias=False), nn.ReLU(inplace=True),
                                 nn.Linear(channels//reduction, channels, bias=False))
        self.sigmoid = nn.Sigmoid()
    def forward(self, x):
        b, c, _, _ = x.shape
        avg_out = self.fc(self.avg_pool(x).view(b, c))
        max_out = self.fc(self.max_pool(x).view(b, c))
        return x * self.sigmoid(avg_out + max_out).view(b, c, 1, 1)

class SpatialAttention(nn.Module):
    def __init__(self, kernel_size=7):
        super().__init__()
        self.conv = nn.Conv2d(2, 1, kernel_size, padding=kernel_size//2, bias=False)
        self.sigmoid = nn.Sigmoid()
    def forward(self, x):
        avg_out = torch.mean(x, dim=1, keepdim=True)
        max_out, _ = torch.max(x, dim=1, keepdim=True)
        return x * self.sigmoid(self.conv(torch.cat([avg_out, max_out], dim=1)))

class CBAM(nn.Module):
    def __init__(self, channels, reduction=16, kernel_size=7):
        super().__init__()
        self.ca, self.sa = ChannelAttention(channels, reduction), SpatialAttention(kernel_size)
    def forward(self, x): return self.sa(self.ca(x))

class DoubleConvCBAM(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
                                   nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True))
        self.cbam = CBAM(out_ch)
    def forward(self, x): return self.cbam(self.conv(x))

class UNet_CBAM(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConvCBAM(in_ch, base), DoubleConvCBAM(base, base*2), DoubleConvCBAM(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvCBAM(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConvCBAM(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConvCBAM(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConvCBAM(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

class SimAM(nn.Module):
    def __init__(self, e_lambda=1e-4):
        super().__init__()
        self.e_lambda = e_lambda
        self.activation = nn.Sigmoid()
    def forward(self, x):
        b, c, h, w = x.shape
        n = w * h - 1
        x_minus_mu_square = (x - x.mean(dim=[2, 3], keepdim=True)).pow(2)
        y = x_minus_mu_square / (4 * (x_minus_mu_square.sum(dim=[2, 3], keepdim=True) / n + self.e_lambda)) + 0.5
        return x * self.activation(y)

class DoubleConvSimAM(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
                                   nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True))
        self.simam = SimAM()
    def forward(self, x): return self.simam(self.conv(x))

class UNet_SimAM(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConvSimAM(in_ch, base), DoubleConvSimAM(base, base*2), DoubleConvSimAM(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvSimAM(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConvSimAM(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConvSimAM(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConvSimAM(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

print("All classes defined. Now load your data/checkpoints as needed.")
'''

os.makedirs("/content/drive/MyDrive/PA_project", exist_ok=True)
with open("/content/drive/MyDrive/PA_project/setup.py", "w") as f:
    f.write(setup_code)

print("setup.py saved to Drive.")
````

````python
def train_and_eval(model_class, name, train_loader, val_loader, test_loader, train_ds, val_ds, num_epochs=30):
    model = model_class(in_ch=1, out_ch=1, base=32).to(device)
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=1e-3)
    for epoch in range(num_epochs):
        model.train()
        train_loss = 0.0
        for sparse, gt in train_loader:
            sparse, gt = sparse.to(device), gt.to(device)
            optimizer.zero_grad()
            output = model(sparse)
            loss = criterion(output, gt)
            loss.backward()
            optimizer.step()
            train_loss += loss.item() * sparse.size(0)
        train_loss /= len(train_ds)
        model.eval()
        val_loss = 0.0
        with torch.no_grad():
            for sparse, gt in val_loader:
                sparse, gt = sparse.to(device), gt.to(device)
                output = model(sparse)
                val_loss += criterion(output, gt).item() * sparse.size(0)
        val_loss /= len(val_ds)
        print(f"[{name}] Epoch {epoch+1}/{num_epochs} | Train: {train_loss:.5f} | Val: {val_loss:.5f}")
    torch.save(model.state_dict(), f"/content/drive/MyDrive/PA_project/checkpoints/{name}.pth")
    model.eval()
    psnr_scores, ssim_scores = [], []
    with torch.no_grad():
        for sparse, gt in test_loader:
            sparse, gt = sparse.to(device), gt.to(device)
            output = model(sparse)
            out_np, gt_np = output.cpu().numpy(), gt.cpu().numpy()
            for i in range(out_np.shape[0]):
                psnr_scores.append(psnr(gt_np[i,0], out_np[i,0], data_range=1.0))
                ssim_scores.append(ssim(gt_np[i,0], out_np[i,0], data_range=1.0))
    p, s = np.mean(psnr_scores), np.mean(ssim_scores)
    print(f"{name} — Test PSNR: {p:.2f} dB | Test SSIM: {s:.4f}\n")
    return model, p, s

print("train_and_eval ready.")
````

````python
import gradio as gr
from skimage.transform import radon, iradon
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim
import numpy as np

def is_likely_vessel_image(gray_array):
    normalized = gray_array / 255.0
    foreground_ratio = np.mean(normalized > 0.5)
    contrast = np.std(normalized)
    reasons = []
    if foreground_ratio > 0.35:
        reasons.append("too much bright area (real vessel images are mostly dark background with thin bright structures)")
    if contrast < 0.08:
        reasons.append("too little contrast (image looks too flat/uniform)")
    is_valid = len(reasons) == 0
    return is_valid, reasons

def process_image(uploaded_image, sparsity_level):
    img = uploaded_image.convert("L").resize((128, 128))
    gt_array = np.array(img).astype(np.float64)

    is_valid, reasons = is_likely_vessel_image(gt_array)
    if not is_valid:
        blank = np.zeros((128, 128))
        warning_msg = (
            "⚠️ **This doesn't look like a vessel-style image.**\n\n"
            f"Reason(s): {', '.join(reasons)}.\n\n"
            "Please upload a binary/high-contrast vessel image (thin white branching structure on a dark background)."
        )
        return blank, blank, blank, "", blank, "", blank, "", blank, "", warning_msg

    gt_norm = gt_array / 255.0
    theta_sparse = np.linspace(0., 180., sparsity_level, endpoint=False)
    sinogram = radon(gt_array, theta=theta_sparse, circle=True)
    sparse_array = iradon(sinogram, theta=theta_sparse, circle=True)
    sparse_array = np.clip(sparse_array, 0, 255)
    sparse_norm = sparse_array / 255.0
    sparse_tensor = torch.from_numpy(sparse_norm).float().unsqueeze(0).unsqueeze(0).to(device)

    results = {}
    for name, model in models_by_sparsity[sparsity_level].items():
        with torch.no_grad():
            out = model(sparse_tensor).cpu().squeeze().numpy()
        out = np.clip(out, 0, 1)
        p = psnr(gt_norm, out, data_range=1.0)
        s = ssim(gt_norm, out, data_range=1.0)
        results[name] = (out, p, s)

    ranked = sorted(results.items(), key=lambda x: x[1][2], reverse=True)
    best_model = ranked[0][0]
    summary = f"**Best result: {best_model}** (highest SSIM at this sparsity level)\n\n"
    summary += "| Model | PSNR | SSIM |\n|---|---|---|\n"
    for name, (_, p, s) in ranked:
        marker = " 🏆" if name == best_model else ""
        summary += f"| {name}{marker} | {p:.2f} dB | {s:.4f} |\n"

    return (
        gt_norm, sparse_norm,
        results["Baseline"][0], f"PSNR: {results['Baseline'][1]:.2f} dB | SSIM: {results['Baseline'][2]:.4f}",
        results["SE"][0], f"PSNR: {results['SE'][1]:.2f} dB | SSIM: {results['SE'][2]:.4f}",
        results["CBAM"][0], f"PSNR: {results['CBAM'][1]:.2f} dB | SSIM: {results['CBAM'][2]:.4f}",
        results["SimAM"][0], f"PSNR: {results['SimAM'][1]:.2f} dB | SSIM: {results['SimAM'][2]:.4f}",
        summary
    )

with gr.Blocks(title="Attention Mechanism Comparison") as demo:
    gr.Markdown("# Attention Mechanism Comparison — Photoacoustic Sparse-View Reconstruction")
    gr.Markdown("Upload a clean vessel image, choose a sparsity level, and compare how each attention mechanism reconstructs it — with live PSNR/SSIM scores.")

    with gr.Row():
        img_input = gr.Image(type="pil", label="Upload a clean vessel image")
        sparsity_input = gr.Radio([32, 64, 96], value=32, label="Sparsity level (sensor angles)")
        run_btn = gr.Button("Run Comparison", variant="primary")

    with gr.Row():
        with gr.Column():
            gt_out = gr.Image(label="Ground Truth (your upload)")
        with gr.Column():
            sparse_out = gr.Image(label="Sparse Input (simulated)")

    gr.Markdown("### Model Outputs")
    with gr.Row():
        with gr.Column():
            base_out = gr.Image(label="Baseline")
            base_score = gr.Markdown()
        with gr.Column():
            se_out = gr.Image(label="SE")
            se_score = gr.Markdown()
        with gr.Column():
            cbam_out = gr.Image(label="CBAM")
            cbam_score = gr.Markdown()
        with gr.Column():
            simam_out = gr.Image(label="SimAM")
            simam_score = gr.Markdown()

    gr.Markdown("### Summary")
    summary_out = gr.Markdown()

    run_btn.click(
        fn=process_image,
        inputs=[img_input, sparsity_input],
        outputs=[gt_out, sparse_out, base_out, base_score, se_out, se_score, cbam_out, cbam_score, simam_out, simam_score, summary_out]
    )

demo.launch(share=True)
````

````python
class SimCBAM(nn.Module):
    """Hybrid: SimAM (parameter-free 'channel-like' attention) + CBAM's spatial attention branch"""
    def __init__(self, channels, kernel_size=7):
        super().__init__()
        self.simam = SimAM()
        self.sa = SpatialAttention(kernel_size)
    def forward(self, x):
        x = self.simam(x)
        x = self.sa(x)
        return x

class DoubleConvSimCBAM(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True)
        )
        self.simcbam = SimCBAM(out_ch)
    def forward(self, x):
        return self.simcbam(self.conv(x))

class UNet_SimCBAM(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConvSimCBAM(in_ch, base), DoubleConvSimCBAM(base, base*2), DoubleConvSimCBAM(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvSimCBAM(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConvSimCBAM(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConvSimCBAM(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConvSimCBAM(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

# Sanity check
model_hybrid = UNet_SimCBAM().to(device)
test_out = model_hybrid(sparse_sample.unsqueeze(0).to(device))
print("SimCBAM U-Net output shape:", test_out.shape)
print("SimCBAM params:", sum(p.numel() for p in model_hybrid.parameters() if p.requires_grad))
````

````python
class SparsityClassifier(nn.Module):
    def __init__(self, num_classes=3):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(1, 16, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(16, 32, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(32, 64, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64 * 16 * 16, 128), nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(128, num_classes)
        )
    def forward(self, x):
        return self.classifier(self.features(x))

print("SparsityClassifier defined.")
````

````python
class SimCBAM(nn.Module):
    def __init__(self, channels, kernel_size=7):
        super().__init__()
        self.simam = SimAM()
        self.sa = SpatialAttention(kernel_size)
    def forward(self, x):
        x = self.simam(x)
        x = self.sa(x)
        return x

class DoubleConvSimCBAM(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True)
        )
        self.simcbam = SimCBAM(out_ch)
    def forward(self, x):
        return self.simcbam(self.conv(x))

class UNet_SimCBAM(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConvSimCBAM(in_ch, base), DoubleConvSimCBAM(base, base*2), DoubleConvSimCBAM(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvSimCBAM(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConvSimCBAM(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConvSimCBAM(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConvSimCBAM(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

print("UNet_SimCBAM redefined.")
````

````python
def auto_process_image(uploaded_image, priority):
    img = uploaded_image.convert("L").resize((128, 128))
    gt_array = np.array(img).astype(np.float64)

    is_valid, reasons = is_likely_vessel_image(gt_array)
    if not is_valid:
        blank = np.zeros((128, 128))
        warning_msg = f"⚠️ This doesn't look like a vessel-style image. {', '.join(reasons)}."
        return blank, blank, "", warning_msg

    gt_norm = gt_array / 255.0

    # Step 1: detect sparsity level using a default 64-angle simulation just for classification
    theta_default = np.linspace(0., 180., 64, endpoint=False)
    sinogram = radon(gt_array, theta=theta_default, circle=True)
    sparse_array_default = np.clip(iradon(sinogram, theta=theta_default, circle=True), 0, 255)

    img_tensor = torch.from_numpy(sparse_array_default / 255.0).float().unsqueeze(0).unsqueeze(0).to(device)
    sparsity_model.eval()
    with torch.no_grad():
        pred = sparsity_model(img_tensor)
        predicted_class = pred.argmax(dim=1).item()
    detected_sparsity = [32, 64, 96][predicted_class]
    confidence = torch.softmax(pred, dim=1)[0][predicted_class].item()

    # Step 2: regenerate the sparse image at the ACTUAL detected sparsity level
    theta_sparse = np.linspace(0., 180., detected_sparsity, endpoint=False)
    sinogram2 = radon(gt_array, theta=theta_sparse, circle=True)
    sparse_array = np.clip(iradon(sinogram2, theta=theta_sparse, circle=True), 0, 255)
    sparse_norm = sparse_array / 255.0
    sparse_tensor = torch.from_numpy(sparse_norm).float().unsqueeze(0).unsqueeze(0).to(device)

    # Step 3: run ALL available models for this sparsity level, LIVE, on this exact image
    live_results = {}
    for name, model in models_by_sparsity[detected_sparsity].items():
        model.eval()
        with torch.no_grad():
            out = model(sparse_tensor).cpu().squeeze().numpy()
        out = np.clip(out, 0, 1)
        p = psnr(gt_norm, out, data_range=1.0)
        s = ssim(gt_norm, out, data_range=1.0)
        # Rough param cost lookup for "efficient" priority
        param_cost = {"Baseline": 0, "SimAM": 0, "SimCBAM": 686, "SE": 13568, "CBAM": 14254}[name]
        live_results[name] = {"output": out, "psnr": p, "ssim": s, "params": param_cost}

    # Step 4: pick the TRUE best model for THIS image, based on chosen priority
    if priority == "psnr":
        best_name = max(live_results, key=lambda k: live_results[k]["psnr"])
    elif priority == "efficient":
        # Best SSIM gain per parameter spent (avoid divide-by-zero for 0-param models by using a small epsilon)
        baseline_ssim = live_results["Baseline"]["ssim"]
        def efficiency_score(k):
            gain = live_results[k]["ssim"] - baseline_ssim
            cost = live_results[k]["params"] + 1  # epsilon to avoid /0
            return gain / cost
        best_name = max(live_results, key=efficiency_score)
    else:  # quality (SSIM)
        best_name = max(live_results, key=lambda k: live_results[k]["ssim"])

    best = live_results[best_name]

    # Build a live leaderboard for transparency
    ranked = sorted(live_results.items(), key=lambda x: x[1]["ssim"], reverse=True)
    leaderboard = "| Model | PSNR | SSIM | Params |\n|---|---|---|---|\n"
    for name, r in ranked:
        marker = " 🏆" if name == best_name else ""
        leaderboard += f"| {name}{marker} | {r['psnr']:.2f} dB | {r['ssim']:.4f} | {r['params']} |\n"

    info = (
        f"**Detected sparsity:** {detected_sparsity} angles (confidence: {confidence:.1%})\n\n"
        f"**Selected model (computed live for THIS image):** {best_name}\n\n"
        f"**Result:** PSNR {best['psnr']:.2f} dB | SSIM {best['ssim']:.4f}\n\n"
        f"### Live leaderboard for this image\n{leaderboard}"
    )

    return sparse_norm, best["output"], "", info

print("auto_process_image upgraded to live, per-image model selection.")
````

````python
setup_code = '''
import os
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from PIL import Image
from torch.utils.data import Dataset, DataLoader
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim
from skimage.transform import radon, iradon
import matplotlib.pyplot as plt

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

if not os.path.exists("/content/vessel_data"):
    os.system("cp -r /content/drive/MyDrive/PA_project/vessel_data /content/vessel_data")

class PADataset(Dataset):
    def __init__(self, sparse_dir, gt_dir):
        self.sparse_dir, self.gt_dir = sparse_dir, gt_dir
        self.files = sorted([f for f in os.listdir(sparse_dir) if f.endswith(".tif")])
    def __len__(self): return len(self.files)
    def __getitem__(self, idx):
        fname = self.files[idx]
        sparse = np.array(Image.open(os.path.join(self.sparse_dir, fname))).astype(np.float32) / 255.0
        gt = np.array(Image.open(os.path.join(self.gt_dir, fname))).astype(np.float32) / 255.0
        return torch.from_numpy(sparse).unsqueeze(0), torch.from_numpy(gt).unsqueeze(0)

# ===== Baseline =====
class DoubleConv(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
                                   nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True))
    def forward(self, x): return self.conv(x)

class UNet(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConv(in_ch, base), DoubleConv(base, base*2), DoubleConv(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConv(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConv(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConv(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConv(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

# ===== SE =====
class SEBlock(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(nn.Linear(channels, channels//reduction, bias=False), nn.ReLU(inplace=True),
                                 nn.Linear(channels//reduction, channels, bias=False), nn.Sigmoid())
    def forward(self, x):
        b, c, _, _ = x.shape
        y = self.fc(self.avg_pool(x).view(b, c)).view(b, c, 1, 1)
        return x * y.expand_as(x)

class DoubleConvSE(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
                                   nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True))
        self.se = SEBlock(out_ch)
    def forward(self, x): return self.se(self.conv(x))

class UNet_SE(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConvSE(in_ch, base), DoubleConvSE(base, base*2), DoubleConvSE(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvSE(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConvSE(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConvSE(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConvSE(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

# ===== CBAM =====
class ChannelAttention(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        self.avg_pool, self.max_pool = nn.AdaptiveAvgPool2d(1), nn.AdaptiveMaxPool2d(1)
        self.fc = nn.Sequential(nn.Linear(channels, channels//reduction, bias=False), nn.ReLU(inplace=True),
                                 nn.Linear(channels//reduction, channels, bias=False))
        self.sigmoid = nn.Sigmoid()
    def forward(self, x):
        b, c, _, _ = x.shape
        avg_out = self.fc(self.avg_pool(x).view(b, c))
        max_out = self.fc(self.max_pool(x).view(b, c))
        return x * self.sigmoid(avg_out + max_out).view(b, c, 1, 1)

class SpatialAttention(nn.Module):
    def __init__(self, kernel_size=7):
        super().__init__()
        self.conv = nn.Conv2d(2, 1, kernel_size, padding=kernel_size//2, bias=False)
        self.sigmoid = nn.Sigmoid()
    def forward(self, x):
        avg_out = torch.mean(x, dim=1, keepdim=True)
        max_out, _ = torch.max(x, dim=1, keepdim=True)
        return x * self.sigmoid(self.conv(torch.cat([avg_out, max_out], dim=1)))

class CBAM(nn.Module):
    def __init__(self, channels, reduction=16, kernel_size=7):
        super().__init__()
        self.ca, self.sa = ChannelAttention(channels, reduction), SpatialAttention(kernel_size)
    def forward(self, x): return self.sa(self.ca(x))

class DoubleConvCBAM(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
                                   nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True))
        self.cbam = CBAM(out_ch)
    def forward(self, x): return self.cbam(self.conv(x))

class UNet_CBAM(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConvCBAM(in_ch, base), DoubleConvCBAM(base, base*2), DoubleConvCBAM(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvCBAM(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConvCBAM(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConvCBAM(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConvCBAM(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

# ===== SimAM =====
class SimAM(nn.Module):
    def __init__(self, e_lambda=1e-4):
        super().__init__()
        self.e_lambda = e_lambda
        self.activation = nn.Sigmoid()
    def forward(self, x):
        b, c, h, w = x.shape
        n = w * h - 1
        x_minus_mu_square = (x - x.mean(dim=[2, 3], keepdim=True)).pow(2)
        y = x_minus_mu_square / (4 * (x_minus_mu_square.sum(dim=[2, 3], keepdim=True) / n + self.e_lambda)) + 0.5
        return x * self.activation(y)

class DoubleConvSimAM(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
                                   nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True))
        self.simam = SimAM()
    def forward(self, x): return self.simam(self.conv(x))

class UNet_SimAM(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConvSimAM(in_ch, base), DoubleConvSimAM(base, base*2), DoubleConvSimAM(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvSimAM(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConvSimAM(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConvSimAM(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConvSimAM(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

# ===== SimCBAM (hybrid, novel) =====
class SimCBAM(nn.Module):
    def __init__(self, channels, kernel_size=7):
        super().__init__()
        self.simam = SimAM()
        self.sa = SpatialAttention(kernel_size)
    def forward(self, x):
        x = self.simam(x)
        x = self.sa(x)
        return x

class DoubleConvSimCBAM(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
                                   nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True))
        self.simcbam = SimCBAM(out_ch)
    def forward(self, x): return self.simcbam(self.conv(x))

class UNet_SimCBAM(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, base=32):
        super().__init__()
        self.enc1, self.enc2, self.enc3 = DoubleConvSimCBAM(in_ch, base), DoubleConvSimCBAM(base, base*2), DoubleConvSimCBAM(base*2, base*4)
        self.pool = nn.MaxPool2d(2)
        self.bottleneck = DoubleConvSimCBAM(base*4, base*8)
        self.up3, self.dec3 = nn.ConvTranspose2d(base*8, base*4, 2, stride=2), DoubleConvSimCBAM(base*8, base*4)
        self.up2, self.dec2 = nn.ConvTranspose2d(base*4, base*2, 2, stride=2), DoubleConvSimCBAM(base*4, base*2)
        self.up1, self.dec1 = nn.ConvTranspose2d(base*2, base, 2, stride=2), DoubleConvSimCBAM(base*2, base)
        self.out_conv = nn.Conv2d(base, out_ch, 1)
    def forward(self, x):
        e1 = self.enc1(x); e2 = self.enc2(self.pool(e1)); e3 = self.enc3(self.pool(e2))
        b = self.bottleneck(self.pool(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.out_conv(d1)

# ===== Sparsity classifier =====
class SparsityClassifier(nn.Module):
    def __init__(self, num_classes=3):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(1, 16, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(16, 32, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(32, 64, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64 * 16 * 16, 128), nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(128, num_classes)
        )
    def forward(self, x):
        return self.classifier(self.features(x))

# ===== train_and_eval =====
def train_and_eval(model_class, name, train_loader, val_loader, test_loader, train_ds, val_ds, num_epochs=30):
    model = model_class(in_ch=1, out_ch=1, base=32).to(device)
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=1e-3)
    for epoch in range(num_epochs):
        model.train()
        train_loss = 0.0
        for sparse, gt in train_loader:
            sparse, gt = sparse.to(device), gt.to(device)
            optimizer.zero_grad()
            output = model(sparse)
            loss = criterion(output, gt)
            loss.backward()
            optimizer.step()
            train_loss += loss.item() * sparse.size(0)
        train_loss /= len(train_ds)
        model.eval()
        val_loss = 0.0
        with torch.no_grad():
            for sparse, gt in val_loader:
                sparse, gt = sparse.to(device), gt.to(device)
                output = model(sparse)
                val_loss += criterion(output, gt).item() * sparse.size(0)
        val_loss /= len(val_ds)
        print(f"[{name}] Epoch {epoch+1}/{num_epochs} | Train: {train_loss:.5f} | Val: {val_loss:.5f}")
    torch.save(model.state_dict(), f"/content/drive/MyDrive/PA_project/checkpoints/{name}.pth")
    model.eval()
    psnr_scores, ssim_scores = [], []
    with torch.no_grad():
        for sparse, gt in test_loader:
            sparse, gt = sparse.to(device), gt.to(device)
            output = model(sparse)
            out_np, gt_np = output.cpu().numpy(), gt.cpu().numpy()
            for i in range(out_np.shape[0]):
                psnr_scores.append(psnr(gt_np[i,0], out_np[i,0], data_range=1.0))
                ssim_scores.append(ssim(gt_np[i,0], out_np[i,0], data_range=1.0))
    p, s = np.mean(psnr_scores), np.mean(ssim_scores)
    print(f"{name} — Test PSNR: {p:.2f} dB | Test SSIM: {s:.4f}\\n")
    return model, p, s

# ===== Vessel image heuristic check =====
def is_likely_vessel_image(gray_array):
    normalized = gray_array / 255.0
    foreground_ratio = np.mean(normalized > 0.5)
    contrast = np.std(normalized)
    reasons = []
    if foreground_ratio > 0.35:
        reasons.append("too much bright area (real vessel images are mostly dark background with thin bright structures)")
    if contrast < 0.08:
        reasons.append("too little contrast (image looks too flat/uniform)")
    is_valid = len(reasons) == 0
    return is_valid, reasons

# ===== Recommendation table (based on 15 full training runs) =====
RECOMMENDATION_TABLE = {
    32: {"best_psnr": "CBAM", "best_psnr_val": 19.02, "best_ssim": "CBAM", "best_ssim_val": 0.8354,
         "most_efficient": "SimCBAM", "efficient_note": "686 params, 94% of CBAM's SSIM gain"},
    64: {"best_psnr": "SE", "best_psnr_val": 21.71, "best_ssim": "CBAM", "best_ssim_val": 0.9015,
         "most_efficient": "SimCBAM", "efficient_note": "686 params, near-SE SSIM performance"},
    96: {"best_psnr": "CBAM", "best_psnr_val": 22.32, "best_ssim": "CBAM", "best_ssim_val": 0.9176,
         "most_efficient": "SimAM", "efficient_note": "0 params, 91.1% SSIM"}
}

def recommend_model(sparsity_level, priority="quality"):
    info = RECOMMENDATION_TABLE[sparsity_level]
    if priority == "psnr":
        return info["best_psnr"], f"Best PSNR at {sparsity_level} angles: {info['best_psnr_val']:.2f} dB"
    elif priority == "efficient":
        return info["most_efficient"], f"Most efficient choice: {info['efficient_note']}"
    else:
        return info["best_ssim"], f"Best SSIM at {sparsity_level} angles: {info['best_ssim_val']:.4f}"

# ===== Checkpoint loading helper =====
def load_model(model_class, path):
    m = model_class().to(device)
    m.load_state_dict(torch.load(path, map_location=device))
    m.eval()
    return m

def load_all_models():
    ckpt_dir = "/content/drive/MyDrive/PA_project/checkpoints"
    models_by_sparsity = {}
    for sparsity in [32, 64, 96]:
        models_by_sparsity[sparsity] = {
            "Baseline": load_model(UNet, f"{ckpt_dir}/baseline_unet_sparse{sparsity}.pth"),
            "SE": load_model(UNet_SE, f"{ckpt_dir}/se_unet_sparse{sparsity}.pth"),
            "CBAM": load_model(UNet_CBAM, f"{ckpt_dir}/cbam_unet_sparse{sparsity}.pth"),
            "SimAM": load_model(UNet_SimAM, f"{ckpt_dir}/simam_unet_sparse{sparsity}.pth"),
            "SimCBAM": load_model(UNet_SimCBAM, f"{ckpt_dir}/simcbam_unet_sparse{sparsity}.pth"),
        }
    return models_by_sparsity

print("All classes, functions, and helpers defined. Call load_all_models() and load the sparsity classifier next.")
'''

os.makedirs("/content/drive/MyDrive/PA_project", exist_ok=True)
with open("/content/drive/MyDrive/PA_project/setup.py", "w") as f:
    f.write(setup_code)

print("setup.py updated and saved to Drive with all 5 architectures + sparsity classifier + demo helpers.")
````

## 2. Methodology

### 2.1 Dataset and Sparse-View Simulation
The dataset used in this work comprises 4,400 vessel images at a $128 \times 128$ resolution, partitioned into training ($n=4,000$), validation ($n=200$), and test ($n=200$) sets. To simulate photoacoustic (PA) tomography acquisition artifacts, ground truth (GT) images were projected into the sinogram domain using the Radon transform. Sparse-view inputs were generated by sampling the sinogram at three discrete sensor densities: 32, 64, and 96 angles uniformly distributed over a 180-degree range. The final artefactual inputs were reconstructed using Filtered Back Projection (FBP) via the Inverse Radon transform, resulting in images characterized by significant streak artifacts and structural discontinuities.

### 2.2 Proposed SimCBAM Attention Module
We propose **SimCBAM**, a lightweight hybrid attention mechanism designed to optimize the trade-off between reconstruction accuracy and parameter efficiency. Standard Convolutional Block Attention Modules (CBAM) utilize a learned Multi-Layer Perceptron (MLP) for channel attention, which introduces significant parameter overhead. SimCBAM replaces this with the parameter-free **SimAM** mechanism, which derives neuron importance via an analytical energy function.

The module retains the spatial attention branch of CBAM, utilizing a $7 \times 7$ convolutional kernel to capture long-range spatial dependencies. Formally, for a feature map $X$, the SimCBAM refinement process is defined as:

$$X_{sim} = \text{SimAM}(X)$$
$$X' = \sigma(f^{7 \times 7}([\text{AvgPool}(X_{sim}); \text{MaxPool}(X_{sim})])) \otimes X_{sim}$$

Where $f^{7 \times 7}$ denotes the convolutional operation and $\sigma$ the sigmoid activation. This hybrid approach reduces the attention-related parameter count from 14,254 (CBAM) to 686 (SimCBAM).

### 2.3 Network Architecture and Training
The backbone architecture is a deep U-Net consisting of three encoder stages, a central bottleneck, and three decoder stages with skip connections. Each block utilizes double $3 \times 3$ convolutions, batch normalization, and ReLU activation. We evaluated five variants: Baseline U-Net, SE U-Net, CBAM U-Net, SimAM U-Net, and our proposed SimCBAM U-Net.

All models were trained for **30 epochs** to ensure full convergence of the loss function. Optimization was performed using the **Adam optimizer** with a fixed learning rate of $1 \times 10^{-3}$ and a **Mean Squared Error (MSE)** loss function. Models were implemented in PyTorch and trained on NVIDIA L4 GPUs.

## 1. Introduction

### 1.1 Background of Photoacoustic Tomography
Photoacoustic (PA) tomography has emerged as a powerful non-invasive imaging modality that synergizes the high contrast of optical imaging with the deep penetration and high spatial resolution of ultrasonic imaging. The underlying physical principle, the photoacoustic effect, involves the absorption of short-duration laser pulses by endogenous chromophores (such as hemoglobin, melanin, or lipids) or exogenous contrast agents. This absorption leads to transient thermoelastic expansion, which subsequently generates ultrasonic waves, termed PA signals. Because ultrasound scatters significantly less than light in biological tissues, PA tomography can achieve sub-millimeter resolution at depths exceeding several centimeters, far surpassing the depth limits of purely optical techniques like confocal microscopy or optical coherence tomography.

### 1.2 The Challenge of Sparse-View Acquisition
In clinical and preclinical settings, high-fidelity PA reconstruction typically requires a dense distribution of ultrasonic transducers around the region of interest to satisfy the Nyquist-Shannon sampling theorem. However, practical implementation often faces significant hardware constraints. These include the high cost of multi-element transducer arrays, limited acoustic access in specific anatomical regions, and the necessity for high-speed acquisition to mitigate motion artifacts in dynamic imaging. Consequently, researchers frequently resort to sparse-view acquisition, where PA signals are collected from a limited number of detector positions or projection angles.

Reconstructing images from such sparse data is a mathematically ill-posed inverse problem. Traditional analytical methods, such as Filtered Back Projection (FBP) or Universal Back Projection (UBP), rely on the assumption of complete data. When applied to sparse-view data, these methods produce reconstructions marred by severe streak artifacts, significant structural degradation, and a low signal-to-noise ratio (SNR). These artifacts often obscure fine vasculature and other critical morphological features, limiting the clinical utility of sparse-view PA systems.

### 1.3 Deep Learning and Attention Mechanisms
Recent advancements in deep learning, particularly the adoption of Convolutional Neural Networks (CNNs), have revolutionized the field of PA image reconstruction. Architectures based on the U-Net have demonstrated remarkable proficiency in post-processing FBP reconstructions to remove artifacts and restore structural connectivity. By learning the non-linear mapping between artefactual sparse-view images and clean ground-truth data, these models effectively act as sophisticated denoising and inpainting filters.

To further enhance the representational power of U-Nets, attention mechanisms have been integrated to allow the network to focus on relevant structural features while suppressing noise and artifacts. The Squeeze-and-Excitation (SE) block and the Convolutional Block Attention Module (CBAM) are two prominent examples. While these modules improve reconstruction fidelity (PSNR and SSIM), they introduce additional learnable parameters. In resource-constrained environments or high-resolution 3D imaging, the cumulative parameter overhead of complex attention blocks can lead to increased memory consumption and computational latency.

### 1.4 Motivation and Proposed Work
There exists a critical need for attention mechanisms that provide the performance benefits of spatial and channel recalibration without the prohibitive computational cost. The Simple Parameter-Free Attention Module (SimAM) offers a potential solution by analytically deriving neuron importance based on an energy function, requiring zero learnable parameters for channel-wise scaling. However, in the context of PA tomography, spatial dependencies are equally crucial for resolving directional streak artifacts.

In this study, we propose **SimCBAM**, a novel hybrid attention module designed specifically for efficient sparse-view PA reconstruction. By integrating the parameter-free SimAM mechanism with the spatial attention branch of CBAM, we hypothesize that a significant portion of reconstruction gain can be achieved at a fraction of the parameter cost. This manuscript details the development of SimCBAM, evaluates its performance against four standard benchmarks across varying sparsity levels (32, 64, and 96 angles), and introduces an automated sparsity-level classifier to enable intelligent, adaptive reconstruction pipelines.

## 2. Related Work

### 2.1 Photoacoustic Vascular Imaging
Photoacoustic (PA) imaging has established itself as a premier modality for visualizing vascular structures due to the strong optical absorption of hemoglobin. The ability to distinguish between oxygenated and deoxygenated hemoglobin allows for high-contrast functional imaging of angiogenesis, which is a critical biomarker in oncology and cardiovascular research. However, capturing these fine, branching structures remains a significant challenge when acquisition views are limited.

### 2.2 Sparse-View Photoacoustic Reconstruction
Traditional reconstruction methods for PA tomography, such as Filtered Back Projection (FBP) and iterative model-based algorithms, are highly sensitive to sampling density. Sparse-view acquisition leads to an incomplete Radon transform, resulting in the characteristic 'streak' artifacts that violate the structural integrity of the vessels. While iterative methods (e.g., Total Variation regularization) can mitigate these effects, they are computationally expensive and often require manual tuning of regularization parameters, making them less suitable for real-time clinical deployment.

### 2.3 Deep Learning in PA Reconstruction
The application of deep learning, particularly the U-Net architecture, has significantly advanced PA reconstruction by treating artifact removal as a supervised image-to-image translation task. U-Nets leverage hierarchical feature extraction and skip connections to preserve high-frequency details while filtering low-frequency artifacts. Recent studies have demonstrated that deep learning can outperform iterative methods in both reconstruction speed and quality under severe sparsity constraints.

### 2.4 Evolution of Attention Mechanisms
Attention mechanisms have been integrated into CNNs to focus computational resources on the most informative parts of a feature map.

*   **Squeeze-and-Excitation (SE):** Introduced by Hu et al., SE blocks recalibrate channel-wise features by explicitly modeling interdependencies between channels. While effective, SE blocks only focus on 'what' is important (channels) and ignore 'where' (spatial location).
*   **Convolutional Block Attention Module (CBAM):** Woo et al. improved upon SE by sequentially applying channel and spatial attention. This dual-pathway approach captures both channel-wise importance and long-range spatial dependencies. However, the channel-attention sub-module in CBAM requires learned Multi-Layer Perceptrons (MLPs), increasing parameter overhead.
*   **Simple Parameter-Free Attention Module (SimAM):** Yang et al. proposed SimAM, which derives 3D weights for feature maps analytically based on an energy function inspired by the mature-neuron theory. SimAM is unique in that it requires zero learnable parameters while providing both channel and spatial recalibration.

### 2.5 Hybrid and Parameter-Efficient Attention
There is an increasing trend toward optimizing attention modules for parameter efficiency. The proposed **SimCBAM** module sits at the intersection of CBAM and SimAM. By replacing the learned channel-attention component of CBAM with the analytical SimAM mechanism, we aim to retain the superior spatial resolving power of CBAM's spatial branch while minimizing the parameter count. This comparison is meaningful as it tests whether learned channel weights are truly necessary for the specific artifacts encountered in sparse-view PA tomography.

## 3. Materials and Methods

### 3.1 Overall Reconstruction Framework
The reconstruction framework is designed as a deep learning-based post-processing pipeline. The objective is to map a low-quality, artefactual image $X_{sparse} \in \mathbb{R}^{H \times W}$, reconstructed via Filtered Back Projection (FBP) from $N$ projection angles, to a high-fidelity estimate $\hat{Y} \in \mathbb{R}^{H \times W}$ that approximates the ground-truth optical absorption map $Y_{gt}$. The framework consists of three modular components: (1) a forward-inverse simulation block for data generation, (2) an attention-enhanced U-Net for artifact suppression, and (3) an automated sparsity classifier for adaptive model deployment.

### 3.2 Dataset and Preprocessing
We utilized a simulated vascular dataset consisting of 4,400 images ($128 \times 128$ pixels) derived from biological vessel patterns. The dataset was partitioned into 4,000 samples for training, 200 for validation, and 200 for independent testing. Preprocessing involved normalizing pixel intensities to the range $[0, 1]$ to stabilize training. To ensure robustness, the images represent varying vascular densities and branching complexities.

### 3.3 Sparse-View Data Generation
To simulate the physical constraints of Photoacoustic (PA) tomography, we implemented a discrete Radon transform to generate sinograms $S$ from GT images:
$$S(\theta, \rho) = \int_{-\infty}^{\infty} \int_{-\infty}^{\infty} f(x,y) \delta(x\cos\theta + y\sin\theta - \rho) dx dy$$
We simulated three levels of acquisition sparsity by sampling $\theta$ at $N \in \{32, 64, 96\}$ uniform intervals over $[0, 180^\circ)$. The resulting sparse sinograms were reconstructed back into the image domain using the Inverse Radon transform with a Ram-Lak filter. This process introduced characteristic streak artifacts and aliasing noise, forming the dataset for supervised learning.

### 3.4 Baseline U-Net Architecture
The core architecture is an 8-layer deep U-Net. The **encoder** pathway utilizes two $3 \times 3$ convolutions per stage, followed by Batch Normalization and ReLU activation, with feature channels doubling at each of the three downsampling steps (32 to 256). The **bottleneck** serves as the latent representation layer. The **decoder** pathway mirrors the encoder, using $2 \times 2$ transposed convolutions to restore spatial resolution, with skip connections concatenating encoder features to preserve high-frequency structural details.

### 3.5 Attention-Enhanced U-Net Architectures
To investigate the performance-to-parameter trade-off, we integrated four distinct attention modules into the double-convolution blocks of the U-Net:

#### 3.5.1 SE-U-Net
Squeeze-and-Excitation (SE) blocks recalibrate channel-wise features. The module "squeezes" spatial information via global average pooling and "excites" channels using a small learnable MLP with a reduction ratio of 16.

#### 3.5.2 CBAM-U-Net
The Convolutional Block Attention Module (CBAM) sequentially applies channel attention (using both average and max pooling) and spatial attention (using a $7 \times 7$ convolution). While powerful, the channel sub-module requires learned weights in the MLP.

#### 3.5.3 SimAM-U-Net
The Simple Parameter-Free Attention Module (SimAM) calculates 3D weights for neurons based on a local energy function. It requires zero learnable parameters, deriving importance from the linear separability of a target neuron from its neighbors.

#### 3.5.4 SimCBAM-U-Net (Proposed Hybrid Module)
We propose **SimCBAM** to address the parameter overhead of CBAM. Our hybrid approach replaces the MLP-based channel attention pathway of CBAM with the parameter-free SimAM pathway, while maintaining the spatial attention branch to ensure directional artifacts are resolved.

### 3.6 Proposed SimCBAM Architecture
The SimCBAM module processes a feature map $F \in \mathbb{R}^{C \times H \times W}$ through two stages:
1.  **Analytical Channel Weighting:** $F_{sim} = \text{SimAM}(F)$, where weights are computed per-neuron based on the energy function minimizing the difference between a neuron and its mean.
2.  **Spatial Refinement:** $M_s(F_{sim}) = \sigma(f^{7 \times 7}([\text{AvgPool}(F_{sim}); \text{MaxPool}(F_{sim})]))$, where the spatial mask is multiplied element-wise by $F_{sim}$.
This configuration ensures that spatial resolving power is maintained at a negligible cost of 686 parameters (from the $7 \times 7$ filter).

### 3.7 Loss Function
Models were optimized using the Mean Squared Error (MSE) loss, which penalizes pixel-wise intensity deviations:
$$\mathcal{L}_{MSE} = \frac{1}{HW} \sum_{i=1}^{H} \sum_{j=1}^{W} (Y_{gt}^{(i,j)} - \hat{Y}^{(i,j)})^2$$

### 3.8 Evaluation Metrics
Performance was evaluated using:
*   **Peak Signal-to-Noise Ratio (PSNR):** $10 \log_{10} (\frac{MAX_I^2}{MSE})$, measuring signal fidelity.
*   **Structural Similarity Index (SSIM):** Comparing luminance, contrast, and structure to measure perceptual quality.
*   **Parameter Count:** Specifically focusing on the additional overhead introduced by attention modules relative to the baseline.

## 5. Experimental Setup

### 5.1 Dataset Configuration
The research utilized a simulated vascular dataset comprising 4,400 biological phantoms representing diverse branching structures and vessel densities. The images, originally sized at $128 \times 128$ pixels, provide a controlled environment to assess reconstruction fidelity. The dataset was partitioned into a large training set ($n=4,000$) to allow the models to learn complex artifact-removal mappings, while validation and testing were performed on independent subsets of 200 images each ($n_{val}=200, n_{test}=200$). All ground-truth images were normalized to a $[0, 1]$ intensity range prior to the simulation of acoustic acquisition.

### 5.2 Sparse-View Acquisition Settings
To emulate hardware-limited photoacoustic tomography (PAT) acquisition, the forward problem was modeled using a discrete Radon transform. We simulated three distinct levels of measurement sparsity to test the robustness of the attention mechanisms under varying degrees of information loss:
*   **32-Angle (Severe Sparsity):** Projections sampled at $5.625^\circ$ intervals over a $180^\circ$ arc.

*   **96-Angle (Low Sparsity):** Projections sampled at $1.875^\circ$ intervals.

In all cases, the initial reconstruction was performed using Filtered Back Projection (FBP) with a Ram-Lak filter. These reconstructions served as the inputs to the U-Net variants, characterized by aliasing streak artifacts and loss of vessel connectivity.

### 5.3 Training Configuration
All models were trained end-to-end using the supervised learning paradigm. The training process was governed by the following hyperparameter settings:
*   **Loss Function:** Mean Squared Error (MSE), optimized to reduce pixel-wise intensity deviations from the ground truth.
*   **Optimizer:** Adam optimizer with an initial learning rate of $1 \times 10^{-3}$, chosen for its adaptive moment estimation capabilities in deep networks.
*   **Epochs:** All models were trained for a duration of **30 epochs**. Preliminary experiments indicated that while 20 epochs provided substantial learning, extending to 30 epochs ensured convergence of the validation loss curves for the more complex attention-enhanced variants.
*   **Batch Size:** A batch size of 16 was utilized to balance gradient stability with memory constraints on the GPU hardware.

### 5.4 Software Environment
The experimental framework was implemented in **Python** using the **PyTorch** deep learning library. Sparse-view simulations (Radon/Inverse Radon transforms) were performed using the **Scikit-Image** library. Training and inference were conducted in a **Google Colab** environment utilizing **NVIDIA L4 Tensor Core GPUs** to facilitate high-speed convolutional operations. ## 7. Software Demonstration and Application Interface

To bridge the gap between algorithmic development and practical deployment, we developed an integrated software interface using the Gradio framework. The application provides two distinct operational modes: a research-oriented manual comparison mode and an end-to-end automated reconstruction pipeline.

### 7.1 Manual-Mode Reconstruction Interface
The manual interface is designed for qualitative and quantitative benchmarking of attention mechanisms. Users can upload a target vascular image and select from three discrete sparsity levels: 32, 64, or 96 projection angles. The system performs a live simulation of the sparse-view acquisition and subsequently generates reconstructions using all five architectural variants (Baseline, SE, CBAM, SimAM, and the proposed SimCBAM). The interface facilitates a side-by-side qualitative comparison of the ground truth, the artefactual sparse input, and the five model outputs, allowing for immediate visual assessment of feature preservation and artifact suppression.

### 7.2 Manual-Mode Performance Summary
For every reconstruction, the application calculates and displays live Peak Signal-to-Noise Ratio (PSNR) and Structural Similarity Index (SSIM) scores. A summary leaderboard is dynamically generated, ranking the models based on SSIM to identify the optimal reconstruction for the specific uploaded image and chosen sparsity level. While these metrics provide valuable individual-sample insights, they are presented as per-image performance data distinct from the aggregate test-set results reported in Section 3.

### 7.3 Automated Reconstruction Mode
The automated mode demonstrates an intelligent imaging workflow that minimizes user intervention. Upon image upload, the system utilizes the trained CNN-based sparsity classifier to automatically detect the sensor density of the input. As established in the experimental results, this classifier operates with 100% accuracy on the independent test set, reliably identifying whether the input corresponds to a 32, 64, or 96-angle acquisition. The detected sparsity level, along with the classification confidence, is reported to the user in real-time.

### 7.4 Automatic Model Selection and Reconstruction
Following sparsity detection, the system triggers the automatic selection of the best-performing reconstruction model according to the detected acquisition parameters. The selected model is applied to the input, and the application displays the final reconstructed vessel image alongside its corresponding PSNR and SSIM scores. This mode serves as a proof-of-concept for an end-to-end clinical workflow where the reconstruction parameters are optimized automatically by the software backend based on raw data characteristics.

### 7.5 Practical Software Workflow
The integrated workflow follows a rigorous sequence: **Input Image → Sparse-View Simulation → CNN-based Sparsity Detection → Intelligent Model Selection → Artifact Removal → Quantitative Evaluation → Output Visualization.** By providing both a granular manual comparison tool and a streamlined automated pipeline, the software demonstrates the technical robustness and practical scalability of the SimCBAM framework for photoacoustic tomographic reconstruction.

### Figure for Introduction: Photoacoustic Forward and Inverse Problem
This cell generates a figure suitable for the introduction of the manuscript. It illustrates the transition from biological tissue to sparse data, and the resulting artifacts that deep learning must solve.

````python
import matplotlib.pyplot as plt
import numpy as np
from skimage.transform import radon, iradon
from PIL import Image

# Load a representative vessel sample
img = np.array(Image.open('/content/vessel_data/Vessel/test/0.tif')).astype(np.float64)
gt_norm = img / 255.0

# Simulate extreme sparsity (32 angles)
theta = np.linspace(0., 180., 32, endpoint=False)
sinogram = radon(gt_norm, theta=theta, circle=True)
sparse_recon = iradon(sinogram, theta=theta, circle=True)

fig, ax = plt.subplots(1, 3, figsize=(18, 6))

ax[0].imshow(gt_norm, cmap='magma')
ax[0].set_title("A: Ground Truth Anatomy\n(Biological Tissue)", fontsize=14, fontweight='bold')
ax[0].axis('off')

ax[1].imshow(sinogram, aspect='auto', cmap='viridis')
ax[1].set_title("B: Sparse Sinogram\n(Acoustic Data @ 32 Angles)", fontsize=14, fontweight='bold')
ax[1].set_xlabel("Detector Index")
ax[1].set_ylabel("Projection Angle")

ax[2].imshow(sparse_recon, cmap='magma')
ax[2].set_title("C: Sparse-View Reconstruction\n(FBP with Streak Artifacts)", fontsize=14, fontweight='bold')
ax[2].axis('off')

plt.tight_layout()
plt.savefig('/content/drive/MyDrive/PA_project/figures/intro_pa_physics.png', dpi=300)
plt.show()
````

````python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

def draw_simcbam_diagram():
    fig, ax = plt.subplots(figsize=(10, 12))
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 12)
    ax.axis('off')

    # Helper to draw boxes
    def draw_box(x, y, text, color='white', edgecolor='black', width=3, height=1):
        rect = patches.FancyBboxPatch((x-width/2, y-height/2), width, height,
                                      boxstyle="round,pad=0.1", linewidth=2,
                                      edgecolor=edgecolor, facecolor=color)
        ax.add_patch(rect)
        ax.text(x, y, text, ha='center', va='center', fontsize=12, fontweight='bold')

    # Blocks
    draw_box(5, 11, "Input Feature Map X", color='#e1f5fe')
    draw_box(5, 9.5, "SimAM Module\n(Parameter-Free)", color='#fff9c4')
    draw_box(5, 8, "F_sim", color='#f5f5f5', edgecolor='gray', width=2)

    draw_box(3.5, 6.5, "AvgPool", color='#c8e6c9', width=2)
    draw_box(6.5, 6.5, "MaxPool", color='#c8e6c9', width=2)

    draw_box(5, 5, "Concatenation", color='#e1f5fe', width=3)
    draw_box(5, 3.5, "7 × 7 Convolution", color='#ffccbc')
    draw_box(5, 2, "Sigmoid (σ)", color='#d1c4e9', width=2)

    ax.text(5, 0.8, "⊗", ha='center', va='center', fontsize=25, fontweight='bold')
    draw_box(5, -0.5, "Output Feature Map X'", color='#e1f5fe')

    # Arrows
    arrow_props = dict(arrowstyle='->', lw=2, color='black')
    ax.annotate('', xy=(5, 10), xytext=(5, 10.5), arrowprops=arrow_props)
    ax.annotate('', xy=(5, 8.5), xytext=(5, 9), arrowprops=arrow_props)

    # Split
    ax.plot([5, 5], [7.5, 8], color='black', lw=2)
    ax.plot([3.5, 6.5], [7.5, 7.5], color='black', lw=2)
    ax.annotate('', xy=(3.5, 7), xytext=(3.5, 7.5), arrowprops=arrow_props)
    ax.annotate('', xy=(6.5, 7), xytext=(6.5, 7.5), arrowprops=arrow_props)

    # Join
    ax.annotate('', xy=(3.5, 5.5), xytext=(3.5, 6), arrowprops=arrow_props)
    ax.annotate('', xy=(6.5, 5.5), xytext=(6.5, 6), arrowprops=arrow_props)
    ax.plot([3.5, 6.5], [5.5, 5.5], color='black', lw=2)
    ax.annotate('', xy=(5, 5), xytext=(5, 5.5), arrowprops=arrow_props)

    ax.annotate('', xy=(5, 4), xytext=(5, 4.5), arrowprops=arrow_props)
    ax.annotate('', xy=(5, 2.5), xytext=(5, 3), arrowprops=arrow_props)
    ax.annotate('', xy=(5, 1.2), xytext=(5, 1.5), arrowprops=arrow_props)

    # Multiplication bypass
    ax.plot([2, 2], [0.8, 8], color='black', lw=2, ls='--')
    ax.plot([2, 4], [8, 8], color='black', lw=2, ls='--')
    ax.annotate('', xy=(4.5, 0.8), xytext=(2, 0.8), arrowprops=arrow_props)

    ax.annotate('', xy=(5, 0), xytext=(5, 0.5), arrowprops=arrow_props)

    plt.title("Architecture of the Proposed SimCBAM Module", fontsize=15, pad=20)
    plt.savefig('/content/drive/MyDrive/PA_project/figures/simcbam_architecture.png', dpi=300, bbox_inches='tight')
    plt.show()

draw_simcbam_diagram()
````
